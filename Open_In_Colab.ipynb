# 📦 Inventory Shortage Tracker — Open in Google Colab

This notebook lets you run the **Inventory Shortage Tracker** web app entirely inside Google Colab — no local install needed.

**What it does:**
1. Installs Node.js inside the Colab environment
2. Clones this repository from GitHub
3. Installs npm dependencies
4. Builds the React app into static files
5. Serves the app and gives you a live clickable URL

▶️ **Just press `Runtime → Run all` (or Shift+Enter through each cell)**

## Step 1 — Install Node.js

```python
# Install Node.js 20 LTS (takes ~30 seconds)
!curl -fsSL https://deb.nodesource.com/setup_20.x | sudo -E bash - > /dev/null 2>&1
!sudo apt-get install -y nodejs > /dev/null 2>&1
!node --version
!npm --version
```

## Step 2 — Clone the repository

```python
import os

REPO_URL = "https://github.com/N1T1N-ANALYST/Bike-Sales-Analysis.git"
REPO_DIR = "/content/Bike-Sales-Analysis"

if os.path.exists(REPO_DIR):
    print("Repository already cloned — pulling latest changes…")
    !git -C {REPO_DIR} pull
else:
    print("Cloning repository…")
    !git clone {REPO_URL} {REPO_DIR}

print("\n✅ Repository ready at", REPO_DIR)
```

## Step 3 — Install npm dependencies

```python
APP_DIR = f"{REPO_DIR}/inventory-shortage-tracker"

print("Installing npm packages (takes ~30 seconds)…")
!cd {APP_DIR} && npm install --silent
print("\n✅ Dependencies installed")
```

## Step 4 — Build the React app

```python
print("Building the app…")
!cd {APP_DIR} && npm run build
print("\n✅ Build complete — static files are in", APP_DIR + "/dist")
```

## Step 5 — Serve the app and get your live URL

The cell below starts a local HTTP server on port **8080** and uses Colab's built-in port proxy to give you a **public URL** you can open in any browser tab.

```python
import subprocess
import threading
import time
from google.colab.output import eval_js

PORT = 8080
DIST_DIR = f"{APP_DIR}/dist"

# Start Python's built-in HTTP server in the background
server = subprocess.Popen(
    ["python3", "-m", "http.server", str(PORT), "--directory", DIST_DIR],
    stdout=subprocess.DEVNULL,
    stderr=subprocess.DEVNULL
)

time.sleep(2)  # Give the server a moment to start

# Get the Colab proxy URL for this port
url = eval_js(f"google.colab.kernel.proxyPort({PORT})")

print("="*60)
print("🚀 Inventory Shortage Tracker is LIVE!")
print("="*60)
print(f"\n👉 Open this URL in a new browser tab:\n\n   {url}\n")
print("💡 Tip: keep this notebook tab open — closing it stops the server.")
print("="*60)
```

## (Optional) Step 6 — Stop the server

Run this cell when you're done to shut down the server.

```python
try:
    server.terminate()
    print("✅ Server stopped.")
except NameError:
    print("Server was not running.")
```
